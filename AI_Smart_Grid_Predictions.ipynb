This notebook demonstrates an AI-based approach to forecast solar energy output and electricity load using machine learning models.

## 🧠 Train AI Models

```python
import pandas as pd
import numpy as np
import joblib
from xgboost import XGBRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split
import os

# Generate synthetic dataset
X = np.random.rand(100, 3)  # Features: [temperature, humidity, wind_speed]
y_solar = X[:, 0] * 100
y_load = X[:, 0] * 50 + X[:, 1] * 20

X_train, X_test, y_train_solar, _ = train_test_split(X, y_solar, test_size=0.2)
_, _, y_train_load, _ = train_test_split(X, y_load, test_size=0.2)

# Train models
solar_model = XGBRegressor()
load_model = RandomForestRegressor()

solar_model.fit(X_train, y_train_solar)
load_model.fit(X_train, y_train_load)

# Save models
os.makedirs("models", exist_ok=True)
joblib.dump(solar_model, "models/solar_model.pkl")
joblib.dump(load_model, "models/load_model.pkl")

print("✅ Models trained and saved.")
```

## 🔮 Forecast Functions

```python
import numpy as np
import joblib

def predict_solar(input_data):
    model = joblib.load("models/solar_model.pkl")
    input_array = np.array(input_data)
    return model.predict(input_array).tolist()

def predict_load(input_data):
    model = joblib.load("models/load_model.pkl")
    input_array = np.array(input_data)
    return model.predict(input_array).tolist()
```

## 📦 Sample Input & Predictions

```python
sample_input = [[0.9, 0.3, 0.2], [0.5, 0.6, 0.4]]

solar_result = predict_solar(sample_input)
load_result = predict_load(sample_input)

print("🔋 Solar Prediction:", solar_result)
print("⚡ Load Prediction:", load_result)
```
